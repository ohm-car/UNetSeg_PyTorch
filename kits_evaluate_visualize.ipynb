```python
def get_model():

    # model = fcn_resnet50(aux_loss=True)
    model = deeplabv3_resnet50(num_classes = KiTS_Dataset.num_classes - 1, aux_loss=True)

    # Access the backbone's first convolutional layer
    original_conv1 = model.backbone.conv1

    # Create a new convolutional layer with 1 input channel
    # Use the same output channels, kernel size, stride, and padding as the original
    new_conv1 = nn.Conv2d(
        in_channels=1, 
        out_channels=original_conv1.out_channels,
        kernel_size=original_conv1.kernel_size,
        stride=original_conv1.stride,
        padding=original_conv1.padding,
        bias=original_conv1.bias
    )

    aux = nn.Sequential(nn.Conv2d(1024, 512, kernel_size=(3, 3), stride=(1, 1), padding=(1, 1), bias=False),
                 nn.BatchNorm2d(512, eps=1e-05, momentum=0.1, affine=True, track_running_stats=True),
                 nn.ReLU(inplace=True),
                 nn.Dropout(p=0.1, inplace=False),
                 nn.Conv2d(512, 1, kernel_size=(1, 1), stride=(1, 1)),
                 nn.Sigmoid())
    model.aux_classifier = aux
    # Replace the original conv1 with the new one
    model.backbone.conv1 = new_conv1
    model.classifier.append(nn.Sigmoid())
    model.to(device=device)
    return model
```

### Visualization Helpers

```python
def gt_torchToPIL_img(img):
    
    img = img.squeeze().cpu().numpy()
#     img = img.transpose((1,2,0))
    return Image.fromarray((img * 255).astype(np.uint8), 'L')
```

```python
def gt_torchToPIL_mask(mask):
    
    mask = mask.squeeze().cpu().numpy()
    mask = np.clip(mask, 0, 1)
    return [Image.fromarray((mask[0] * 255).astype(np.uint8), 'L'), Image.fromarray((mask[1] * 255).astype(np.uint8), 'L')]
```

```python
def ml_torchToPIL_mask(mask):
    
    mask = mask.squeeze().cpu().detach().numpy()
    mask = np.clip(mask, 0, 1)
    mask[0] += mask[1]
    return [Image.fromarray((mask[0] * 255).astype(np.uint8), 'L'), Image.fromarray((mask[1] * 255).astype(np.uint8), 'L')]
```

```python
def seg_torchToPIL_mask(mask):
    
    mask = mask.squeeze().cpu().detach().numpy()
    mask = np.clip(np.round(mask), 0, 1)
    mask[0] += mask[1]
    return [Image.fromarray((mask[0] * 255).astype(np.uint8), 'L'), Image.fromarray((mask[1] * 255).astype(np.uint8), 'L')]
```

```python
def combined_display_pic(images):
    
#     widths, heights = 224, 224
#     total_width = sum(widths)
#     max_height = max(heights)

    widths, heights = 512, 512
    total_width = widths*len(images) + 10*(len(images) - 1)
    max_height = heights
    
    white_spacing = Image.new('RGB', (10, 224), color='white')

    combined_image = Image.new('RGB', (total_width, max_height))
    x_offset = 0
    for img in images:
        combined_image.paste(img, (x_offset, 0))
        x_offset += img.width
        combined_image.paste(white_spacing, (x_offset, 0))
        x_offset += 10

    return combined_image
```

```python
def single_iou(pred_mask, mask):
    
    iou = binary_jaccard_index(pred_mask, mask)
    
    return iou
```

```python
def random_pred_expected_iou(perc):
    
    e_iou = perc / (2 - perc)
    
    return e_iou
```
